```python
# basic operations
# add
a = np.array([1,2,3,4])
b = np.array([5,6,7,8])
print(a+b)

```

```python
# sub
a = np.array([1,2,3,4])
b = np.array([5,6,7,8])
print(a-b)
```

```python
# multiply
a = np.array([1,2,3,4])
b = np.array([5,6,7,8])
print(a*b)
```

```python
# div
a = np.array([1,2,3,4])
b = np.array([5,6,7,8])
print(a/b)
```

```python
# exponetial
a = np.array([1,2,3,4])
b = np.array([5,6,7,8])
print(a**2)
print(b>=6)
np.sqrt(a)
np.exp(a)
np.add(a,b)
print(a.max())
print(b.min())
print(a.cumsum())
```
